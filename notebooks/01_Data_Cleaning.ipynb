## Top 10 Customers by Sales
this chart shows the customers who generated the highest sales.

```python
top_customers = df.groupby("Customer Name")["Sales"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))
top_customers.plot(kind="bar")

plt.title("Top 10 Customers by Sales")
plt.xlabel("Customer Name")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

## Monthly Sales Trend Analysis

```python
import matplotlib.pyplot as plt
# Create Month-Year column
df['Month-Year'] = df['Order Date'].dt.to_period('M')

# Monthly Sales
monthly_sales = df.groupby('Month-Year')['Sales'].sum()

plt.figure(figsize=(12,5))
monthly_sales.plot(kind='line', marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month-Year')
plt.ylabel('Sales')
plt.xticks(rotation=45)

plt.show()
```

## Top 10 Products by Sales

```python
# Top 10 Products by Sales
 
top_products = df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6))
top_products.plot(kind="bar", color="orange")
plt.title("Top 10 Products by Sales")
plt.xlabel("Product Name")
plt.ylabel("Total Sales")
plt.xticks(rotation=75)

plt.show()
```

### Insight 

- The top selling products contribute a significant portion of total revenue.
- These products can be prioritize for inventory management and marketing campaigns.

## Profit by Category

```python
profit_by_category = df.groupby("Category")["Profit"].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))

profit_by_category.plot(kind="bar", color=["green", "orange", "red"])

plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)
plt.show()
```

### Insight

- Technology generates the highest profit.
- Furniture and Office Supplies contribute comparatively lower profit.
- Business should focus on high-profit categories for better growth.
